```python
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)
```

```python
# Create a 4x7 matrix with random numbers in [-0.5, 0.5]
# This represents word embeddings for 4 words, each with dimension 7
X0 = np.random.uniform(-0.5, 0.5, size=(4, 7))

print("Part (a): Word Embeddings Matrix X0")
print("Shape:", X0.shape)
print(X0)
print("\n" + "="*80 + "\n")
```

```python
# Create positional embeddings using the actual formula from transformer papers
# For position pos and dimension i:
# PE(pos, 2i) = sin(pos / 10000^(2i/d_model))
# PE(pos, 2i+1) = cos(pos / 10000^(2i/d_model))

def create_positional_encoding(num_positions, d_model):
    """
    Create positional encodings for transformer
    num_positions: number of positions (words) in sequence
    d_model: dimension of embeddings
    """
    P = np.zeros((num_positions, d_model))

    for pos in range(num_positions):
        for i in range(0, d_model, 2):
            # Calculate the denominator for the sinusoidal functions
            denominator = np.power(10000, (2 * i) / d_model)

            # Apply sin to even indices
            P[pos, i] = np.sin(pos / denominator)

            # Apply cos to odd indices (if within bounds)
            if i + 1 < d_model:
                P[pos, i + 1] = np.cos(pos / denominator)

    return P

# Create positional embeddings for 4 words with dimension 7
P = create_positional_encoding(4, 7)

print("Part (b): Positional Embeddings Matrix P")
print("Shape:", P.shape)
print(P)
print("\n" + "="*80 + "\n")
```

```python
# Following slide 20: p + x vectors
# Add word embeddings X0 and positional embeddings P element-wise
X = X0 + P

print("Part (c): Combined Matrix X = X0 + P")
print("Shape:", X.shape)
print(X)
print("\n" + "="*80 + "\n")
```

```python
# From slide 20: Before computing attention, we project to higher dimension
# Original dimension e = 7, projecting to e' = 13
# We need three weight matrices to create Q, K, V

e = 7   # Original embedding dimension
e_prime = 13  # Projected dimension

# Create three weight matrices with random values in [-0.5, 0.5]
# W^Q: maps from dimension 7 to dimension 13
W_Q = np.random.uniform(-0.5, 0.5, size=(e, e_prime))

# W^K: maps from dimension 7 to dimension 13
W_K = np.random.uniform(-0.5, 0.5, size=(e, e_prime))

# W^V: maps from dimension 7 to dimension 13
W_V = np.random.uniform(-0.5, 0.5, size=(e, e_prime))

print("Part (d): Weight Matrices for Query, Key, Value")
print("\nW^Q (Query weights) - Shape:", W_Q.shape)
print(W_Q)
print("\nW^K (Key weights) - Shape:", W_K.shape)
print(W_K)
print("\nW^V (Value weights) - Shape:", W_V.shape)
print(W_V)
print("\n" + "="*80 + "\n")
```

```python
# From slide 14 and 20: Q = XW^Q, K = XW^K, V = XW^V
# Following slide 10: We treat X as having words in rows (4 words × 7 dimensions)
# Matrix multiplication: (4×7) × (7×13) = (4×13)

Q = X @ W_Q  # Query matrix
K = X @ W_K  # Key matrix
V = X @ W_V  # Value matrix

print("Part (e): Query, Key, and Value Matrices")
print("\nQuery Matrix Q - Shape:", Q.shape)
print(Q)
print("\nKey Matrix K - Shape:", K.shape)
print(K)
print("\nValue Matrix V - Shape:", V.shape)
print(V)
print("\n" + "="*80 + "\n")
```

```python
# From slides 14-15 and 21:
# Head = Attention(Q, K, V) = [SoftMax(QK^T / sqrt(e'))]V

def softmax(x):
    """
    Compute softmax along the last axis
    Subtracting max for numerical stability
    """
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)

# Step 1: Compute QK^T (from slide 10: matrix multiplication)
# Q is (4×13), K^T is (13×4), result is (4×4)
QK_T = Q @ K.T

print("Part (f): Computing Attention-Weighted Output (Head)")
print("\nStep 1: QK^T - Shape:", QK_T.shape)
print(QK_T)

# Step 2: Divide by sqrt(e') for scaled dot-product attention
# From slide 14: this is the "scaled" part
scaling_factor = np.sqrt(e_prime)
scaled_scores = QK_T / scaling_factor

print("\nStep 2: QK^T / sqrt(e') where sqrt(e')=", scaling_factor)
print("Scaled Attention Scores - Shape:", scaled_scores.shape)
print(scaled_scores)

# Step 3: Apply softmax row-wise (from slide 11 and 15)
# Each row represents attention scores for one query word
attention_weights = softmax(scaled_scores)

print("\nStep 3: SoftMax(QK^T / sqrt(e'))")
print("Attention Weights - Shape:", attention_weights.shape)
print(attention_weights)
print("\nNote: Each row sums to 1.0, representing attention distribution")
print("Row sums:", np.sum(attention_weights, axis=1))

# Step 4: Multiply attention weights by V (from slide 15 and 21)
# attention_weights is (4×4), V is (4×13), result is (4×13)
# This weights each value vector by the attention scores
Head_output = attention_weights @ V

print("\nStep 4: [SoftMax(QK^T / sqrt(e'))] × V")
print("Head Output (Attention-Weighted Input) - Shape:", Head_output.shape)
print(Head_output)

print("\n" + "="*80)
print("FINAL RESULT: Single-Head Attention Output")
print("="*80)
print("Shape:", Head_output.shape)
print(Head_output)
print("\nThis matrix contains the attention-weighted representation of the 4 words,")
print("where each word (row) is now represented as a weighted combination of all words")
print("based on their relevance/attention scores.")
```
